```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("ecommerce-medallion-pipeline") \
    .getOrCreate()

# print("Done")
```

```python
LAKEHOUSE_ABFS_PATH = "abfss://ecommerce_medallion_pipeline@onelake.dfs.fabric.microsoft.com/ecommerce_medallion_pipeline_lakehouse.Lakehouse/Files/raw_data"
LAKEHOUSE_AWS_ABFS_PATH = "abfss://ecommerce_medallion_pipeline@onelake.dfs.fabric.microsoft.com/ecommerce_medallion_pipeline_lakehouse.Lakehouse/Files/aws-s3-logistics-data"
# print("Done")
```

```python
# departments
df_departments = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_ABFS_PATH}/departments/")
df_departments.write.format("delta").mode("overwrite").saveAsTable("bronze_departments")
# print("Departments done.")

# employees
df_employees = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_ABFS_PATH}/employees/")
df_employees.write.format("delta").mode("overwrite").saveAsTable("bronze_employees")
# print("Employees done.")

# orders
df_orders = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_ABFS_PATH}/orders/")
df_orders.write.format("delta").mode("overwrite").saveAsTable("bronze_orders")
# print("Orders done.")

# order details
df_order_details = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_ABFS_PATH}/order_details/")
df_order_details.write.format("delta").mode("overwrite").saveAsTable("bronze_order_details")
# print("Order Details done.")

# order details
df_categories = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_ABFS_PATH}/categories/")
df_categories.write.format("delta").mode("overwrite").saveAsTable("bronze_categories")
# print("Order Details done.")

# shippers
df_shippers = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_ABFS_PATH}/shippers/")
df_shippers.write.format("delta").mode("overwrite").saveAsTable("bronze_shippers")
# print("shippers done.")

# product_suppliers
df_product_suppliers = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_ABFS_PATH}/product_suppliers/")
df_product_suppliers.write.format("delta").mode("overwrite").saveAsTable("bronze_product_suppliers")
# print("product_suppliers done.")

# orders_cdc_incremental
df_orders_cdc_incremental = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_ABFS_PATH}/orders_cdc_incremental/")
df_orders_cdc_incremental.write.format("delta").mode("overwrite").saveAsTable("bronze_orders_cdc_incremental")
# print("orders_cdc_incremental done.")
```

```python
df_customers = spark.read.format("csv").option("header", "true").option("inferSchema", "true").option("sep", ";").option("inferSchema", "true").load(f"{LAKEHOUSE_ABFS_PATH}/customers/")
df_customers.write.format("delta").mode("overwrite").saveAsTable("bronze_customers")
# print("Customers done.")

df_products = spark.read.format("csv").option("header", "true").option("inferSchema", "true").option("inferSchema", "true").option("escape", '"').load(f"{LAKEHOUSE_ABFS_PATH}/products/")
df_products.write.format("delta").mode("overwrite").saveAsTable("bronze_products")
# print("Products done.")
```

```python
# payments from s3
df_payments = spark.read.format("csv").option("header", "true").option("inferSchema", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_AWS_ABFS_PATH}/payments/")
df_payments.write.format("delta").mode("overwrite").saveAsTable("bronze_payments")
# print("Payments done.")

# shipments from s3
df_shipments = spark.read.format("csv").option("header", "true").option("inferSchema", "true").option("inferSchema", "true").load(f"{LAKEHOUSE_AWS_ABFS_PATH}/shipments/")
df_shipments.write.format("delta").mode("overwrite").saveAsTable("bronze_shipments")
# print("Shipments done.")
```
